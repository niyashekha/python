```python
#print first and last elements from a list
colors=input("Enter the colors:").split(",")
print("First color :",colors[0])
print("Last color :",colors[-1])
```

```python
#print extension of file from filename
filename=input("Enter the filename:")
extension=filename.split(".")[-1]
print("Extension:",extension)
```

```python
#sum of all elements from list
num=input("Enter the numbers :").split(",")
int_num=list(map(int,num))
print("List:",int_num)
print("Sum:",sum(int_num))
```
