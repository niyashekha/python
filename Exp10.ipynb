```python
#count the occurence of each word in a line of text
line=input("Enter a line of text: ")
words=line.split(",")
count={}
for word in words:
    if word in count:
        count[word]+=1
    else:
        count[word]=1
print("word occurences:")
for word in count:
    print(word,":",count[word])
    
```

```python
#replace values greater than 100 with 'over'
num=list(map(int,input("Enter the integers:").split(",")))
for i in range(len(num)):
    if num[i]>100:
        num[i]="over"
print("New List:",num)
```

```python
#count the occurences of 'a' within a list of first names
name=input("Enter the names:").split(",")
count=0
for n in name:
    count=count + n.lower().count('a')
print("Number of occurences of 'a':",count)
```

```python
#find gcd of two numbers
n1=int(input("Enter first number:"))
n2=int(input("Enter second number"))
while n2!=0:
    r=n1%n2
    n1=n2
    n2=r
print("GCD:",n1)
```

```python
#create a list removing even numbers
num=list(map(int,input("Enter integers:").split(",")))
result=[]
for n in num:
    if n%2 !=0:
        result.append(n)
print("New List:",result)
```

```python
#factorial of a number
n=int(input("Enter a number:"))
fact=1
for i in range(1,n+1):
    fact=fact*i
print("factorial =",fact)
```

```python
#Fibonacci series of n terms
n=int(input("Enter number of terms:"))
a=0
b=1
for i in range(n):
    print(a,end=" ")
    c=a+b
    a=b
    b=c
```

```python
#find sum of all item in a list
l=list(map(int,input("Enter integers:").split(",")))
sum=0
for n in l:
    sum=sum+n
print("Sum =",sum)
```

```python
import math
start=int(input("Enter start:"))
end=int(input("Enter end: "))
for n in range(start,end+1):
    if n>=1000 and n<=9999:
        r=math.isqrt(n)
        if r*r==n:
            s=str(n)
            if all(int(d)%2==0 for d in s):
                print(n)
```

```python
#count character frequency in a string
s=input("Enter a string:")
count={}
for char in s:
        if char in count:
            count[char]+=1
        else:
            count[char]=1
for char in count:
    print(char,":",count[char])
```

```python
#compare two list
l1=list(map(int,input("Enter first list:").split(",")))
l2=list(map(int,input("Enter second list:").split(",")))
if len(l1)==len(l2):
    print("Same length")
else:
    print("Different length")
if sum(l1)==sum(l2):
    print("Same sum")
else:
    print("Different sum")
if set(l1)&set(l2):
    print("Common Values exists")
else:
    print("No common values")
```
