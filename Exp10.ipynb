```python
#count the occurence of each word in a line of text
line=input("Enter a line of text: ")
words=line.split()
count={}
for word in words:
    if word in count:
        count[word]+=1
    else:
        count[word]=1
print("word occurences:")
for word in count:
    print(word,":",count[word])
    
```

```python
#replace values greater than 100 with 'over'
num=list(map(int,input("Enter the integers:").split()))
for i in range(len(num)):
    if num[i]>100:
        num[i]="over"
print("New List:",num)
```

```python
#count the occurences of 'a' within a list of first names
n=input("Enter the names:").split()
count=0
for name in n:
    for char in name:
        if char=='a':
            count+=1
print("Number of occurences of 'a':",count)
```

```python
#find gcd of two numbers
n1=int(input("Enter first number:"))
n2=int(input("Enter second number"))
while n2!=0:
    r=n1%n2
    n1=n2
    n2=r
print("GCD:",n1)
```

```python
#create a list removing even numbers
num=list(map(int,input("Enter integers:").split()))
result=[]
for n in num:
    if n%2 !=0:
        result.append(n)
print("New List:",result)
```

```python
#factorial of a number
n=int(input("Enter a number:"))
fact=1
for i in range(1,n+1):
    fact=fact*i
print("factorial =",fact)
```

```python
#Fibonacci series of n terms
n=int(input("Enter number of terms"))
a=0
b=1
while n>0:
    print(a,end=" ")
    c=a+b
    a=b
    b=c
    n=n-1
```

```python
#find sum of all item in a list
l=list(map(int,input("Enter integers:").split()))
sum=0
for n in l:
    sum=sum+n
print("Sum =",sum)
```

```python
#count character frequency in a string
s=input("Enter a string:")
count={}
for char in s:
        if char in count:
            count[char]+=1
        else:
            count[char]=1
for char in count:
    print(char,":",count[char])
```

```python
#compare two list
l1=list(map(int,input("Enter first list:").split()))
l2=list(map(int,input("Enter second list:").split()))
if len(l1)==len(l2):
    print("Same length")
else:
    print("Different length")
s1=0
s2=0
for x in l1:
    s1+=x
for x in l2:
    s2+=x
if s1==s2:
    print("Same sum")
else:
    print("Different sum")
com=False
for x in l1:
    if x in l2:
        com=True
        break
if com:
    print("Common exists")
else:
    print("no common value")
```
