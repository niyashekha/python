```python
#add 'ing' to the end of string or'ly' if it already ends with 'ing'
s=input("Enter a String:")
if s[-3:]=="ing":
    s=s+"ly"
else:
    s=s+"ing"
print(s)
```

```python
#Check string is palindrome or not
s=input("Enter a String:")
if s==s[::-1]:
    print(s,"is Palindrome")
else:
    print(s," is not Palindrome")
```
