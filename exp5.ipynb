```python
#print 3rd element and 3rd element from last of tuple
num=num=tuple(input("Enter the numbers :").split(","))
print("3rd element :",num[2])
print("3rd element from last :",num[-3])
```

```python
#input from user and print average from tuple
num=input("Enter the numbers :").split(",")
int_num=tuple(map(int,num))
print("Tuple:",int_num)
print("Average :",sum(int_num)/len(num))
```
