```python
#create graphics package and its subpackage
import os
os.makedirs("graphics/graphics3D",exist_ok=True)
```

```python
%%writefile graphics/rectangle.py
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
%%writefile graphics/circle.py
import math 
def area(r):
    return math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
%%writefile graphics/graphics3D/cuboid.py
def area(l,b,h):
    return 2*(l*b + b*h +h*l)
def perimeter(l,b,h):
    return 4*(l+b+h)
```

```python
%%writefile graphics/graphics3D/sphere.py
import math
def sarea(r):
    return 4*math.pi*r*r
```

```python
#Main program
from graphics import rectangle, circle
from graphics.graphics3D import *

while True:
    print("\n1.Rectangle")
    print("2.Circle")
    print("3.Cuboid")
    print("4.Sphere")
    print("5.Exit")

    ch = int(input("Enter your choice: "))

    if ch == 1:
        l = int(input("Enter the length of Rectangle:"))
        b = int(input("Enter the breadth of Rectangle:"))
        print("Rectangle Area:", rectangle.area(l,b))
        print("Rectangle Perimeter:", rectangle.perimeter(l,b))

    elif ch == 2:
        r = int(input("Enter the radius of circle:"))
        print("Circle Area:", round(circle.area(r), 2))
        print("Circle Perimeter:", round(circle.perimeter(r), 2))

    elif ch == 3:
        l = int(input("Enter the length of Cuboid:"))
        b = int(input("Enter the breadth of Cuboid:"))
        h = int(input("Enter the height of Cuboid:"))
        print("Cuboid Surface Area:", cuboid.area(l,b,h))
        print("Cuboid Total Edge Length:", cuboid.perimeter(l,b,h))

    elif ch == 4:
        r = int(input("Enter radius of sphere:"))
        print("Sphere Surface Area:", round(sphere.sarea(r),2))

    elif ch == 5:
        print("Exiting the program...")
        break

    else:
        print("Invalid choice! Please enter 1 to 5.")
```
