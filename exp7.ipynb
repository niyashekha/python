```python
#print colors in set1 Not in set 2
s1=set(input("Enter colors for set 1(separated by commas):").split(","))
s2=set(input("Enter colors for set 2(separated by commas):").split(","))
print("set 1 : ",s1)
print("set 2 : ",s2)
print("Colors in set 1 not in set 2 : ",s1-s2)
```

```python
#print colors in set1 and set 2
s1=set(input("Enter colors for set 1(separated by commas):").split(","))
s2=set(input("Enter colors for set 2(separated by commas):").split(","))
print("set 1 : ",s1)
print("set 2 : ",s2)
print("Colors in set1 and set2 : ",s1|s2)
```
