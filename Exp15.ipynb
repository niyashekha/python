```python
#create rectangle class to calculate area and perimeter and compare two rectangle objects
class rectangle:
    def get(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def Area(self):
        return self.length * self.breadth
    def Perimeter(self):
        return 2*(self.length + self.breadth)
l1=int(input("Enter the Length of Rectangle1 : "))
b1=int(input("Enter the Breadth of Rectangle1 : "))
l2=int(input("Enter the Length of Rectangle2 : "))
b2=int(input("Enter the Breadth of Rectangle2 : "))
r1=rectangle()
r2=rectangle()
r1.get(l1,b1)
r2.get(l2,b2)
print("\nRectangle 1")
print("Area:",r1.Area())
print("Perimeter:",r1.Perimeter())
print("\nRectangle 2")
print("Area:",r2.Area())
print("Perimeter:",r2.Perimeter())

if r1.Area()>r2.Area():
    print("\nRectangle 1 has greater Area")
elif r1.Area()<r2.Area():
    print("\nRectangle 2 has greater Area")
else:
    print("\nBoth Rectangles have Same Area")
```

```python
#Bank account using class and objects
class bank:
    def __init__(self,name,accno,acctype,balance):
        self.name=name
        self.accno=accno
        self.acctype=acctype
        self.balance=balance
    def deposit(self,amount):
        print("Initial Balance:",self.balance)
        self.balance+=amount
        print("Current Balance:",self.balance)
    def withdraw(self):
        amount=int(input("\nEnter withdrawal amount:"))
        if amount<=self.balance:
            self.balance-=amount
            print("Current Balance:",self.balance)
        else:
            print("Not enough Balance")
            
    def display(self):
        print("\nAccount holder name:",self.name)
        print("Account number:",self.accno)
        print("Account type:",self.acctype)
        print("Balance:",self.balance)
name=input("Enter account holder name:")
accno=int(input("Enter Account Number:"))
acctype=input("Enter Account Type:")
balance=int(input("Enter initial balance"))
b=bank(name,accno,acctype,balance)
while True:
    print("\n1.Deposit")
    print("2.Withdraw")
    print("3.Account Info")
    print("4.Exit")
    ch=int(input("Enter your choice:"))
    if ch==1:
        amount=int(input("\nEnter Amount to deposit: "))
        b.deposit(amount)
    elif ch==2:
        b.withdraw()
    elif ch==3:
        b.display()
    elif ch==4:
        print("Exiting...")
        break
        
        
        
```

```python
#implement multilevel inheritance using classes with constructor ivocation and method overriding
class Publisher:
    def __init__(self,name):
        self.name=name
    def display(self):
        print("Publisher name:",self.name)
class Book(Publisher):
    def __init__(self,name,title,author):
        super().__init__(name)
        self.title=title
        self.author=author
    def display(self):
        super().display()
        print("Book title:",self.title)
        print("Author:",self.author)
class Python(Book):
    def __init__(self,name,title,author,price,pages):
        super().__init__(name,title,author)
        self.price=price
        self.pages=pages
    def display(self):
        print("\nBook details:")
        super().display()
        print("Price:",self.price)
        print("Number of pages:",self.pages)
name=input("Enter publisher name: ")
title=input("Enter book title: ")
author=input("Enter author name: ")
price=input("Enter price: ")
pages=input("Enter number of pages: ")
p=Python(name,title,author,price,pages)
p.display()

        
      
        
```

```python
#operator overloading to compare rectangle areas(__gt__)
class Rectangle:
    def __init__(self,l,b):
        self.length=l
        self.breadth=b
        self.a=l*b
    def __gt__(self,other):
        if self.a>other.a:
            print("\nRectangle 1 is greater")
        elif self.a<other.a:
            print("\nRectangle 2 is greater")
        else:
            print("\nBoth rectangles are equal")
l1=int(input("Enter the Length of Rectangle1 : "))
b1=int(input("Enter the Breadth of Rectangle1 : "))
l2=int(input("Enter the Length of Rectangle2 : "))
b2=int(input("Enter the Breadth of Rectangle2 : "))
r1=Rectangle(l1,b1)
r2=Rectangle(l2,b2)
r1>r2
        
    
```

```python
#time addition using operator overloading(+)
class Time:
    def __init__(self,hr,mint,sec):
        self.hour=hr
        self.minute=mint
        self.sec=sec
    def __add__(self,other):
        hh=self.hour+other.hour
        mm=self.minute+other.minute
        ss=self.sec+other.sec
        if ss>=60:
            m=int(ss/60)
            ss = ss%60
            mm+=m
        if mm>=60:
            h =int(mm/60)
            mm = mm%60
            hh+=h
        print("\nAdded time..")
        print("Hour:",hh,"Minute:",mm,"Second:",ss)
h1=int(input("Enter hour1: "))
m1=int(input("Enter minute1: "))
s1=int(input("Enter second1: "))
h2=int(input("Enter hour2: "))
m2=int(input("Enter minute2: "))
s2=int(input("Enter second2: "))
t1=Time(h1,m1,s1)
t2=Time(h2,m2,s2)
t1+t2
```
