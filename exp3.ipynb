```python
#First and last character interchange
s=input("Enter a String : ")
n=s[-1]+s[1:-1]+s[0]
print("New String :",n)
```

```python
#first 2 and last 2 chars
s=input("Enter a string : ")
n=s[:2]+s[-2:]
print("New String :",n)
```

```python
#swap first character in two strings
s1=input("Enter a String")
s2=input("Enter a String")
n1=s2[0]+s1[1:]
n2=s1[0]+s2[1:]
print("New String :",n1 + " " + n2)
```

```python
#Replace all occurence of first character with $
s=input("Enter a string : ")
f=s[0]
n=f+s[1:].replace(f,"$")
print("New String :",n)
```
