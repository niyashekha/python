```python
#Generate postive list of numbers from given integers list
n=list(map(int,input("Enter integers:").split(",")))
p=[x for x in n if x>0]
print(p)
```

```python
#Square of N numbers
n=int(input("Enter the limit:"))
s=[x**2 for x in range(1,n+1)]
print(s)
```

```python
#Form a list of vowels from a given word
word=input("Enter a word:")
vowels=[ch for ch in word if ch in "aeiou"]
print(vowels)
```

```python
#ordinal value of each character in a word
word=input("Enter a word:")
o=[ord(ch) for ch in word]
print(o)
```
