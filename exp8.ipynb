```python
#sort student dictionary by  roll number
roll=list(map(int,input("Enter the Roll Number of 5 Students:").split(",")))
name=input("Enter the Name of 5 Students:").split(",")
student=dict(zip(roll,name))
print("Dictionary:",student)
asec=dict(sorted(student.items()))
print("Ascending :",asec)
desc=dict(sorted(student.items(),reverse=True))
print("Descending :",desc)
```

```python
#merge dictionaries
roll=list(map(int,input("Enter the Roll Number of 3 Students:").split(",")))
name=input("Enter the Name of 3 Students:").split(",")
d1=dict(zip(roll,name))
print("Dictionary 1:",d1)
roll=list(map(int,input("Enter the Roll Number of 2 Students:").split(",")))
name=input("Enter the Name of 2 Students:").split(",")
d2=dict(zip(roll,name))
print("Dictionary 2:",d2)
d1.update(d2)
print("Merged Dictionary:",d1)
```
